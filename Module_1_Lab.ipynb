## 3. Python virtual environments (`venv`)

A **virtual environment** gives a project its own private set of installed packages, so
projects don't break each other.

In Colab, creating a venv does **not** switch the notebook's kernel. So we call the venv's
own Python explicitly: `/content/module1-venv/bin/python`. No activation needed.

````python
# Create the venv. (If this Colab image lacks ensurepip, fall back to virtualenv.)
!python -m venv /content/module1-venv || (pip install -q virtualenv && virtualenv /content/module1-venv)
````

````python
!/content/module1-venv/bin/python -m pip install -q rich
````

````python
!/content/module1-venv/bin/python -c "import sys, rich; print('venv Python:', sys.executable); print('Rich installed successfully')"
````

Compare `venv Python` above with `Notebook Python` from Topic 2 — they are different interpreters.

## 4. Writing and running a Python program

`%%writefile <path>` on the **first line** of a cell saves the rest of the cell to a file
instead of running it.

````python
%%writefile /content/hello.py
from rich import print


def greet(name):
    return f"Hello, {name}!"


if __name__ == "__main__":
    print(greet("World"))
````

Run it with the virtual environment's Python:

````python
!/content/module1-venv/bin/python /content/hello.py
````

Expected output:

```
Hello, World!
```

The file lives on Colab's temporary machine; this notebook holds the instructions to recreate it.

## 5. Unit testing with `unittest`

A **test** is code that checks other code automatically. Each `test_...` method asserts an expected result.

````python
%%writefile /content/test_hello.py
import unittest

from hello import greet


class TestGreeting(unittest.TestCase):
    def test_world(self):
        self.assertEqual(greet("World"), "Hello, World!")

    def test_another_name(self):
        self.assertEqual(greet("Colab"), "Hello, Colab!")


if __name__ == "__main__":
    unittest.main()
````

````python
!cd /content && /content/module1-venv/bin/python test_hello.py
````

You should see `Ran 2 tests` and `OK`. If a test fails, fix the code and rerun **before** submitting.

## 6. Recording dependencies (`requirements.txt`)

`pip freeze` lists the exact package versions installed in the venv. Saving them lets anyone
reinstall the same dependencies later with `pip install -r requirements.txt`.

````python
!/content/module1-venv/bin/python -m pip freeze > /content/requirements.txt
!cat /content/requirements.txt
````

> Never upload the `module1-venv` folder itself — only `requirements.txt`.
